```python
import pandas as pd
```

```python
# 1. Carregar os dados corretamente (como mostrado anteriormente)
sarapui2010 = pd.read_csv('C:/Users/dania/Documents/chestpain/weather-sarapui/dataset/sarapui - 2010.csv', decimal=',', na_values=['-9.999,00'])

sarapui2020 = pd.read_csv('C:/Users/dania/Documents/chestpain/weather-sarapui/dataset/sarapui - 2020.csv', decimal=',', na_values=['-9999,00'])

sarapui2014 = pd.read_csv('C:/Users/dania/Documents/chestpain/weather-sarapui/dataset/sarapui - 2014.csv', decimal=',', na_values=['-9.999,00'])

sarapui2024 = pd.read_csv('C:/Users/dania/Documents/chestpain/weather-sarapui/dataset/sarapui - 2024.csv', decimal=',', na_values=['-9999,00'])
```

```python
# 2. Converter a coluna de temperatura para numérico
sarapui2010['Temperatura ºC'] = pd.to_numeric(
    sarapui2010['Temperatura ºC'],
    errors='coerce'  # Converte valores inválidos para NaN
)
print(sarapui2010.head())

sarapui2020['Temperatura ºC'] = pd.to_numeric(
    sarapui2020['Temperatura ºC'],
    errors='coerce'  # Converte valores inválidos para NaN
)

sarapui2024['Temperatura ºC'] = pd.to_numeric(
    sarapui2024['Temperatura ºC'],
    errors='coerce'  # Converte valores inválidos para NaN
)
```

```python
#2010
maior_temp2010 = sarapui2010['Temperatura ºC'].max()
menor_temp2010 = sarapui2010['Temperatura ºC'].min()

#2020
maior_temp2020 = sarapui2020['Temperatura ºC'].max()
menor_temp2020 = sarapui2020['Temperatura ºC'].min()

#2014
maior_temp2014 = sarapui2014['Temperatura ºC'].max()
menor_temp2014 = sarapui2014['Temperatura ºC'].min()

#2024
maior_temp2024 = sarapui2024['Temperatura ºC'].max()
menor_temp2024 = sarapui2024['Temperatura ºC'].min()
```

```python
print("----------------2010----------------")
# Filtrar as linhas onde ocorreram a maior e menor temperatura
dia_maior_temp2010 = sarapui2010[sarapui2010['Temperatura ºC'] == maior_temp2010]
dia_menor_temp2010 = sarapui2010[sarapui2010['Temperatura ºC'] == menor_temp2010]

print("\nDia(s) com a MAIOR temperatura:")
print(dia_maior_temp2010[['Data UTC', 'Hora BRT', 'Temperatura ºC']])

print("\nDia(s) com a MENOR temperatura:")
print(dia_menor_temp2010[['Data UTC', 'Hora BRT', 'Temperatura ºC']])

print("----------------2020----------------")

dia_maior_temp2020 = sarapui2020[sarapui2020['Temperatura ºC'] == maior_temp2020]
dia_menor_temp2020 = sarapui2020[sarapui2020['Temperatura ºC'] == menor_temp2020]

print("\nDia(s) com a MAIOR temperatura:")
print(dia_maior_temp2020[['Data UTC', 'Hora BRT', 'Temperatura ºC']])

print("\nDia(s) com a MENOR temperatura:")
print(dia_menor_temp2020[['Data UTC', 'Hora BRT', 'Temperatura ºC']])

print("----------------2014----------------")

dia_maior_temp2014 = sarapui2014[sarapui2014['Temperatura ºC'] == maior_temp2014]
dia_menor_temp2014 = sarapui2014[sarapui2014['Temperatura ºC'] == menor_temp2014]

print("\nDia(s) com a MAIOR temperatura:")
print(dia_maior_temp2014[['Data UTC', 'Hora BRT', 'Temperatura ºC']])

print("\nDia(s) com a MENOR temperatura:")
print(dia_menor_temp2014[['Data UTC', 'Hora BRT', 'Temperatura ºC']])

print("----------------2024----------------")

dia_maior_temp2024 = sarapui2024[sarapui2024['Temperatura ºC'] == maior_temp2024]
dia_menor_temp2024 = sarapui2024[sarapui2024['Temperatura ºC'] == menor_temp2024]

print("\nDia(s) com a MAIOR temperatura:")
print(dia_maior_temp2024[['Data UTC', 'Hora BRT', 'Temperatura ºC']])

print("\nDia(s) com a MENOR temperatura:")
print(dia_menor_temp2024[['Data UTC', 'Hora BRT', 'Temperatura ºC']])
```
